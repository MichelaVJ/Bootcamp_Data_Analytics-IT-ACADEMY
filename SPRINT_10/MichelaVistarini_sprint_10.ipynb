```python
import pandas as pd

df = pd.read_excel("sprint10.xlsx", header= 3, index_col= 0)

print("\nColumnes del DF:")
print(list(df.columns))
print("\nDimensió del DF:")
print(df.shape)

df_ordenat = df.sort_values(
    by= ["País d'origen", "Ciutat"], 
    ascending= True)


print("\nEl camp DNI només conté valors únics:", df["DNI"].is_unique)

print("\nDF ordenat per País d'origen i Ciutat, primeres 10 files")

display(df_ordenat.head(10))
```

<div style="color: white;">

Crea una columna que sigui el nom complet.

Crea una columna si la persona és nascuda a Espanya o no.

Posa el DNI com a índex del DataFrame, és a dir, com a noms de files.

Substitueix el nom de les columnes `Dia de Naixement`, `Mes de Naixement` i `Any de Naixement` per `Dia`, `Mes` i `Any`.

Substitueix `H` per `Home`, `D` per `Dona`, `A` per `Altres` i `NC` per una dada faltant: `NaN`, `NULL` o `NA`.

Mostra tots els canvis que has realitzat en una sola taula.

```python
df["Nom complet"] = df["Nom"] + " " + df["Cognoms"]

columna = df.pop("Nom complet")
df.insert(0, "Nom complet", columna)

df["Nascuda a Espanya"] = df["País d\'origen"] == "Espanya"

if "DNI" in df.columns:
    df = df.set_index("DNI")

df = df.rename(
    columns={
        "Dia de Naixement": "Dia",
        "Mes de Naixement": "Mes",
        "Any de Naixement": "Any"
    }
)

df["Gènere"] = df["Gènere"].replace(
    {
        "H": "Home",
        "D": "Dona",
        "A": "Altres",
        "NC": pd.NA
    }
)
display(df.head(15))
```

<div style="color: white;">

Junta les columnes `Fills` i `No Fills` en una sola columna, utilitzant el mètode `.apply()` i definint una funció que resolgui el problema.

La columna nova ha de dir-se `Fills` i prendre els valors `Sí` o `No`.

</div>


```python
def classificar_fills(fila):
    if pd.notna(fila["Fills"]):
        return "Si"
    elif pd.notna(fila["No Fills"]):
        return "No"
    else:
        return pd.NA

df["Fills"] = df.apply(classificar_fills, axis=1)

df = df.drop(columns=["No Fills"])

display(df.head(10))
```

<div style="color: white;">

Crea una taula resum que permeti veure el sou mig, medià, mínim i màxim per `Gènere`.

Ordena la taula en funció del sou mig.

</div>



Modifiquem el tipus de dades de la columna Salari Mensual, passem de text a número per poder operar.

```python
df.select_dtypes(include="number")

df["Salari mensual"] = (
    df["Salari mensual"]
    .astype(str)
    .str.replace("€", "", regex=False)
    .str.replace(".", "", regex=False)
    .str.replace(",", ".", regex=False)
    .str.strip()
)

df["Salari mensual"] = pd.to_numeric(df["Salari mensual"], errors="coerce")

df = df.rename(columns={"Salari mensual": "Salari mensual €"})
```

```python
display(df)
```

Ara podem crear la taula resum amb el sous per gènere

```python
taula_resum = df.groupby("Gènere")["Salari mensual €"].agg(
    sou_mig="mean",
    sou_medià="median",
    sou_mínim="min",
    sou_màxim="max"
)

taula_resum = taula_resum.sort_values(
    by="sou_mig",
    ascending=False
).round(2)

display(taula_resum)
```

<div style="color: white;">

Crea una taula resum amb el salari mig per `Gènere` com a files i `País d'origen` com a columnes.

Afegeix-hi les mitjanes als marges de la taula.

**Extra:** aplica format condicional a la taula per veure amb un color més intens els valors més elevats.

</div>



```python
taula_salari = pd.pivot_table(
    df,
    values="Salari mensual €",
    index="Gènere",
    columns="País d'origen",
    aggfunc="mean",
    margins=True,
    margins_name="Mitjana total"
)

columnes_ordenades = (
    taula_salari
    .loc["Mitjana total"]
    .drop("Mitjana total")
    .sort_values(ascending=False)
    .index
    .tolist()
)

columnes_ordenades.append("Mitjana total")

taula_salari = taula_salari[columnes_ordenades]

display(
    taula_salari
    .style
    .background_gradient(
        cmap="Greens",
        subset=pd.IndexSlice[
            taula_salari.index.drop("Mitjana total"),
            taula_salari.columns.drop("Mitjana total")
        ]
    )
    .format("{:.0f} €", na_rep="-")
)
```

<div style="color: white;">

Crea una columna nova que sigui la data de naixement en format `Datetime` a partir de les columnes `Dia`, `Mes` i `Any`.

Utilitzant aquesta columna, crea una funció que, donada una data, calculi l'edat actual a dia d'avui.

Utilitza la funció que acabes de crear per generar una columna nova al DataFrame amb l'edat actual.

</div>

```python
df["Data naixement"] = pd.to_datetime(
    dict(
        year=df["Any"],
        month=df["Mes"],
        day=df["Dia"]
    ),
    errors="coerce"
)

df = df.drop(columns=["Dia", "Mes", "Any"])

columna_data = df.pop("Data naixement")
df.insert(6, "Data naixement", columna_data)

```

```python
def calcular_edat(data_naixement):
    
    if pd.isna(data_naixement):
        return pd.NA
    
    avui = pd.Timestamp.today()
    
    edat = avui.year - data_naixement.year
    
    if (avui.month, avui.day) < (data_naixement.month, data_naixement.day):
        edat -= 1
    
    return edat

df["Edat actual"] = df["Data naixement"].apply(calcular_edat)

columna_edat = df.pop("Edat actual")
df.insert(7, "Edat actual", columna_edat)

display(df)
```

Utilitzant el següent DataFrame, adjunta la columna `Increment` al DataFrame del nivell anterior.

Actualitza la columna `Salari` en funció dels percentatges que s'adjunten.

No modifiquis manualment els increments: escriu codi Python per fer les conversions necessàries.

</div>


```python
df_increment = pd.DataFrame({
    "Grup": ["Grup A", "Grup B", "Grup C", "Grup D"],
    "Increment": ["5%", "3,5%", "2%", "8%"]
})

display(df_increment)
```

```python
df["Grup Professional"] = df["Grup Professional"].str.strip()
df_increment["Grup"] = df_increment["Grup"].str.strip()

mapa_increments = df_increment.set_index("Grup")["Increment"]

df["Increment"] = df["Grup Professional"].map(mapa_increments)

columna_increment = df.pop("Increment")
df.insert(11, "Increment", columna_increment)

display(df)
```

```python
def convertir_percentatge(valor):
    valor = valor.replace("%", "")
    valor = valor.replace(",", ".")
    valor = float(valor) / 100
    return valor

df["Increment decimal"] = df["Increment"].apply(convertir_percentatge)

df["Salari mensual €"] = (df["Salari mensual €"] * (1 + df["Increment decimal"])).round(2)

display(df)
```

<div style="color: white;">

Utilitzant un bucle, exporta en 4 fitxers, en format `.xlsx` o `.csv`, les dades de cada `Grup Professional`.

Per exemple:

- `dades_GrupA.xlsx`
- `dades_GrupB.xlsx`
- `dades_GrupC.xlsx`
- `dades_GrupD.xlsx`

Exporta un 5è DataFrame, en format `.xlsx` o `.csv`, que contingui:

- quants treballadors hi ha per cada `Grup Professional`;
- quin és el seu sou mig;
- quina és la seva edat mediana.

</div>

```python
from pathlib import Path

carpeta_sortida = Path("exportacions_grups")
carpeta_sortida.mkdir(exist_ok=True)

for grup, df_grup in df.groupby("Grup Professional"):
    
    nom_grup = grup.replace(" ", "")

    nom_fitxer = f"dades_{nom_grup}.xlsx"
    
    df_grup.to_excel(carpeta_sortida / nom_fitxer, index=True)
```

```python
resum_grups = (
    df.groupby("Grup Professional")
    .agg(
        Treballadors=("Grup Professional", "count"),
        Sou_mig=("Salari mensual €", "mean"),
        Edat_mediana=("Edat actual", "median")
))

resum_grups["Sou_mig"] = resum_grups["Sou_mig"].round(2)
resum_grups["Edat_mediana"] = resum_grups["Edat_mediana"].round(0).astype(int)


display(resum_grups)

resum_grups.to_excel(carpeta_sortida / "resum_grups.xlsx", index=True)
```
