Des d'un Jupyter Notebook faràs els següents exercicis utilitzant la llibreria `requests` de Python.
Exploració bàsica amb una API de laboratori.

* 1. Consulta l'API pública [JSONPlaceholder](https://jsonplaceholder.typicode.com/) utilitzant el mètode GET per obtenir:
   * Llista de publicacions (`/posts`)
   * Llista d'usuaris (`/users`)
   * Llista de tasques (`/todos`)

* 2. Mostra per pantalla:
   * La quantitat total de cada recurs.
   * El codi d'estat de cada petició.

```python
import requests
import pandas as pd
```

```python
BASE = 'https://jsonplaceholder.typicode.com'
recursos = ['posts', 'users', 'todos']

for recurs in recursos:
    r = requests.get(f'{BASE}/{recurs}')
    dades = r.json()
    print(f'/{recurs} → codi: {r.status_code} | total: {len(dades)}')
```

* 3. Fes una petició a una publicació inexistent per obtenir un error 404 i mostra el codi d'estat rebut.

```python
r = requests.get(f'{BASE}/posts/9999')
print(f'Codi d\'estat: {r.status_code}')
print(f'Resposta: {r.json()}')
```

* 4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.
   * Mostra la resposta JSON.
   * Mostra el codi d'estat.

```python
nova_publicacio = {
    'title': 'IT Academy - Barcelona Activa',
    'body': "Aquest és l'últim sprint del bootcamp",
    'userId': 1
}

r = requests.post(f'{BASE}/posts', json=nova_publicacio)
print(f'Codi d\'estat: {r.status_code}')
print(f'Resposta JSON: {r.json()}')
```

* 5. Fes una petició PATCH per modificar parcialment una publicació existent.
   * Mostra la resposta JSON.
   * Mostra el codi d'estat.

<span style="color: #0099FF;">Visualitzem el post 1, per saber que volem canviar</span>

```python
r = requests.get(f'{BASE}/posts/1')
print(r.json())
```

<span style="color: #0099FF;">Canviem el títol</span>

```python
canvis = {
    'title': 'Títol actualitzat amb PATCH'
}

r = requests.patch(f'{BASE}/posts/1', json=canvis)
print(f'Codi d\'estat: {r.status_code}')
print(f'Resposta JSON: {r.json()}')
```

* 6. Fes una petició DELETE sobre una publicació.
   * Mostra la resposta JSON.
   * Mostra el codi d'estat.

```python
r = requests.delete(f'{BASE}/posts/1')
print(f'Codi d\'estat: {r.status_code}')
print(f'Resposta JSON: {r.json()}')
```

* 3. Fes una petició GET senzilla:
   * Mostra el codi d'estat de la resposta.
   * Imprimeix de forma clara alguns dels camps de la resposta JSON.

```python
r = requests.get('https://catfact.ninja/facts?limit=10')
print(f'Codi d\'estat: {r.status_code}')

dades = r.json()
facts = dades['data']

print(f'\nTotal de random facts: {len(facts)}')
print('\nAlguns exemples:')
for f in facts[:3]:
    print(f'  - {f["fact"]} ({f["length"]} caràcters)')
```

# <span style="color: #FF00FF;">Nivell 3</span>

API d'[Open Data Barcelona](https://opendata-ajuntament.barcelona.cat/ca/desenvolupadors#APIS)

* 1. Utilitza l'API d'Open Data BCN per:
   * Cercar un dataset del teu interès mitjançant `package_search` i `package_show`.

* 2. Dels resultats obtinguts, selecciona'n un que tingui recursos disponibles en CSV o JSON.

```python
BASE_BCN = 'https://opendata-ajuntament.barcelona.cat/data/api/action'

r = requests.get(f'{BASE_BCN}/package_search', params={'q': 'medi ambient', 'rows': 10})
print(f'Codi d\'estat: {r.status_code}')

resultats = r.json()['result']['results']
for ds in resultats:
    print(f'  - {ds["name"]} → {ds["title"]}')
```

<span style="color: #0099FF;">Comprovem si el dataset té recursos CSV o JASON </span>

```python
r = requests.get(f'{BASE_BCN}/package_show', params={'id': 'atles-biodiversitat-arbrat-especies'})
detalls = r.json()['result']

for rec in detalls['resources']:
    print(f'  - {rec["name"]} | format: {rec["format"]} | datastore_active: {rec["datastore_active"]}')
```

<span style="color: #0099FF;">Busquem l'id del recurs que volem descarregar </span>

```python
for rec in detalls['resources']:
    if rec['datastore_active']:
        print(f'resource_id: {rec["id"]}')
```

* 3. Amb el `resource_id` del recurs seleccionat, realitza una consulta mitjançant `datastore_search` per:
   * Recuperar almenys 100 registres.

```python
r = requests.get(f'{BASE_BCN}/datastore_search', params={
    'resource_id': 'c654c488-2822-4e51-ad9b-f443989a4b41',
    'limit': 100
})
print(f'Codi d\'estat: {r.status_code}')

dades = r.json()['result']
print(f'Registres obtinguts: {len(dades["records"])}')
print(f'Total disponible: {dades["total"]}')
```

* 4. Converteix els resultats a un DataFrame.

```python
df_arbres = pd.DataFrame(dades['records'])
print(f'Shape: {df_arbres.shape}')
df_arbres.head()
```


* 5. Desa el DataFrame en un fitxer `.csv`.


```python
df_arbres.to_csv('arbres_barcelona.csv', index=False, encoding='utf-8')
print('Fitxer desat: arbres_barcelona.csv')
```
