**1. Calculadora de l'índex de massa corporal**

Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en: [🔗 Índice de masa corporal IMC que es y como se calcula.](https://muysalud.com/salud/indice-de-masa-corporal-imc-que-es-y-como-se-calcula/)

La funció ha de classificar el resultat en les seves respectives categories.

*Consell: Intenta validar les dades prèviament, perquè enviï un missatge d'advertència si les dades introduïdes per l'usuari/a no estan en el format adequat o no pren valors raonables.*


```python
def calcular_imc():
    try:
        pes = float(input("Introdueix el teu pes en kg (ex. 64): "))
        alcada = float(input("Introdueix la teva alçada en metres (ex. 1.75): "))

        if pes <= 0 or alcada <= 0:
            print("Error: el pes i l'alçada han de ser valors positius.")
        else:
            imc = pes / (alcada ** 2)

            print("IMC:", round(imc, 2))

            if imc < 18.5:
                print("Categoria: Pes baix")
            elif imc < 25:
                print("Categoria: Pes normal")
            elif imc < 30:
                print("Categoria: Sobrepès")
            else:
                print("Categoria: Obesitat")

    except ValueError:
        print("Error: has d'introduir números.")
```

**2. Convertidor de temperatures**

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).

Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera que es puguin guardar (recorda que un `print()` no es pot guardar mai).

*Consell: Intenta validar les dades prèviament, perquè enviï un missatge d'advertència si les dades introduïdes per l'usuari/a no estan en el format adequat.*

**(EXTRA):** Pensa una manera d'emmagatzemar totes les possibles conversions en un sol objecte (Llista? Diccionari? DataFrame?) en comptes d'escriure molts `if else` en funció de la temperatura d'origen i la temperatura de destí.


```python
def convertir_temperatura():

    # Creem diccionari amb possibles conversions
    conversions = {
        "C": {
            "F": lambda temperatura: (temperatura * 9/5) + 32,
            "K": lambda temperatura: temperatura + 273.15
        },
        "F": {
            "C": lambda temperatura: (temperatura - 32) * 5/9,
            "K": lambda temperatura: (temperatura - 32) * 5/9 + 273.15
        },
        "K": {
            "C": lambda temperatura: temperatura - 273.15,
            "F": lambda temperatura: (temperatura - 273.15) * 9/5 + 32
        }
    }

    # Temperatures mínimes
    limits_minims = {
        "C": -273.15,
        "F": -459.67,
        "K": 0
    }

    # Demanem la temperatura a l'usuari
    temperatura_text = input("Introdueix la temperatura: ")

    # Validem que la temperatura sigui un número
    try:
        temperatura = float(temperatura_text)
    except ValueError:
        return "Avís: la temperatura ha de ser un número."

    # Demanem la unitat d'origen
    unitat_origen = input("Introdueix la unitat d'origen: C, F o K: ").upper()

    # Validem que la unitat introduïda existeixi
    if unitat_origen not in conversions:
        return "Avís: la unitat ha de ser C, F o K."

    # Validem que la temperatura no sigui inferior al zero absolut
    if temperatura < limits_minims[unitat_origen]:
        return "Avís: la temperatura introduïda no és físicament possible."

    # Guardem els resultats en un diccionari
    resultat = {}

    # Recorrem totes les conversions disponibles per a la unitat d'origen
    for unitat_desti, formula in conversions[unitat_origen].items():
        resultat[unitat_desti] = round(formula(temperatura), 2)

    return resultat
```

```python
resultat_conversio = convertir_temperatura()

print(resultat_conversio)
```

**3. Comptador de paraules d'un text**

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula. Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament.

**(EXTRA):** Quina és la llargada mitjana de les paraules del text que has escrit? "Hola com va?" hauria de retornar (4+3+2) / 3 = 3


```python
def comptar_paraules(text):

    # Validem text, ha de ser tipus string
    if not isinstance(text, str):
        return "Avís: has d'introduir un text."

    # Eliminem espais
    text = text.strip()

    # Validem que el text no estigui buit
    if text == "":
        return "Avís: el text no pot estar buit."

    # Passem tot el text a minúscules
    text = text.lower()

    # Eliminem signes de puntuació
    signes_puntuacio = ".,;:!?¿¡()[]{}"

    for signe in signes_puntuacio:
        text = text.replace(signe, "")

    # Separem el text en paraules
    paraules = text.split()

    # Validem que hi hagi almenys una paraula
    if len(paraules) == 0:
        return "Avís: el text ha de contenir almenys una paraula."

    # Creem un diccionari per guardar el recompte
    recompte_paraules = {}

    # Recorrem cada paraula del text
    for paraula in paraules:

        # Si la paraula ja existeix al diccionari, sumem 1
        if paraula in recompte_paraules:
            recompte_paraules[paraula] += 1

        # Si la paraula no existeix, la guardem amb valor 1
        else:
            recompte_paraules[paraula] = 1

    # Calculem la llargada total de totes les paraules
    llargada_total = 0

    for paraula in paraules:
        llargada_total += len(paraula)

    # Calculem la llargada mitjana
    llargada_mitjana = round(llargada_total / len(paraules), 2)

    resultat = "Recompte de paraules:\n"

    for paraula, quantitat in recompte_paraules.items():
        resultat += f"{paraula}: {quantitat}\n"

    resultat += f"\nLlargada mitjana de les paraules: {llargada_mitjana}"

    return resultat
```

```python
text_usuari = input("Introdueix un text: ")

resultat = comptar_paraules(text_usuari)

print(f"Frase introduïda: {text_usuari}")
print(resultat)
```

**4. Diccionari invers (amb possibilitat de duplicats)**

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors.

Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.


```python
def invertir_diccionari(diccionari):

    # Validem que el que rebem sigui un diccionari
    if type(diccionari) != dict:
        return "Avís: has d'introduir un diccionari."

    # Validem que el diccionari no estigui buit
    if len(diccionari) == 0:
        return "Avís: el diccionari no pot estar buit."

    # Creem un diccionari buit per guardar el resultat
    diccionari_invertit = {}

    # Recorrem el diccionari original
    for clau, valor in diccionari.items():

        # Si el valor encara no existeix al diccionari invertit
        if valor not in diccionari_invertit:
            diccionari_invertit[valor] = clau

        # Si el valor ja existeix, vol dir que hi ha un duplicat
        else:
            print(f"Avís: el valor '{valor}' està repetit.")

            # Si encara no és una llista, el convertim en llista
            if type(diccionari_invertit[valor]) != list:
                diccionari_invertit[valor] = [diccionari_invertit[valor]]

            # Afegim la nova clau a la llista
            diccionari_invertit[valor].append(clau)

            print(f"Claus associades al valor repetit: {diccionari_invertit[valor]}")

    return diccionari_invertit
```

**1. Comptador i endreçador de paraules d'un text**

El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z. Per exemple, per a l'arxiu "tu\_me\_quieres\_blanca.txt".


```python
def comptar_i_ordenar_paraules(nom_arxiu):

    # Obrir arxiu
    try:
        arxiu = open(nom_arxiu, "r", encoding="utf-8")
        text = arxiu.read()
        arxiu.close()

    except FileNotFoundError:
        return "Avís: no s'ha trobat l'arxiu."

    # Eliminem espais del principi i del final
    text = text.strip()

    # Validem que l'arxiu no estigui buit
    if text == "":
        return "Avís: l'arxiu està buit."

    # Passem tot el text a minúscules
    text = text.lower()

    # Eliminem signes de puntuació
    signes_puntuacio = ".,;:!?¿¡()[]{}«»\"'"

    for signe in signes_puntuacio:
        text = text.replace(signe, "")

    # Separem el text en paraules
    paraules = text.split()

    # Diccionari on guardarem el recompte general de paraules
    recompte_paraules = {}

    # Comptem quantes vegades apareix cada paraula
    for paraula in paraules:

        if paraula in recompte_paraules:
            recompte_paraules[paraula] += 1
        else:
            recompte_paraules[paraula] = 1

    # Diccionari final ordenat per lletra inicial
    resultat = {}

    # Abecedari per l'ordre de la A a la Z
    abecedari = "abcdefghijklmnopqrstuvwxyz"

    # Recorrem cada lletra de l'abecedari
    for lletra in abecedari:

        # Diccionari temporal per guardar les paraules que comencen per aquesta lletra
        paraules_lletra = {}

        # Recorrem les paraules ordenades alfabèticament
        for paraula in sorted(recompte_paraules):

            # Comprovem si la paraula comença per la lletra actual
            if paraula.startswith(lletra):
                paraules_lletra[paraula] = recompte_paraules[paraula]

        # Si hi ha paraules amb aquesta lletra, les afegim al resultat
        if len(paraules_lletra) > 0:
            resultat[lletra] = paraules_lletra

    return resultat
```

**2. Conversió de tipus de dades**

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client:


```python
def conversio(llista_dades):
    convertibles = []
    no_convertibles = []

    def revisar_element(element):
        if type(element) == list or type(element) == tuple:
            for valor in element:
                revisar_element(valor)
        else:
            try:
                numero = float(element)
                convertibles.append(numero)
            except:
                no_convertibles.append(element)

    for element in llista_dades:
        revisar_element(element)

    return convertibles, no_convertibles
```

**1. Generador de Contrasenyes**

Explora el funcionament del mòdul random de la llibreria numpy. [🔗 Random module in NumPy](https://www.cybrosys.com/blog/random-module-in-numpy)

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d'aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:

- `longitud (int)`: Longitud de la contrasenya
- `majuscules (bool = True)`: Si hi ha d'aparèixer majúscules
- `minuscules (bool = True)`: Si hi ha d'aparèixer minúscules
- `numeros (bool = True)`: Si hi ha d'aparèixer números
- `signes (bool = False)`: Si hi ha d'aparèixer caràcters especials (,-$? o similars)


Així doncs, si executem la funció de la següent manera: `crear_contrasenya(10, True, True, True, True)`

Hauríem d'obtenir un output (que hem de poder guardar) de l'estil: `9Er,5Vn8P$`

Assegura't que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.

**(EXTRA)** Explora com podríem fer que la funció copiés la contrasenya automàticament al porta papers de l'ordinador (com si l'haguéssim seleccionat i fet ctrl+copy).


```python
import numpy as np

def crear_contrasenya(longitud, majuscules=True, minuscules=True, numeros=True, signes=False):
    grup_majuscules = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    grup_minuscules = "abcdefghijklmnopqrstuvwxyz"
    grup_numeros = "0123456789"
    grup_signes = ".,-$?!@#%&"

    grups_activats = []

    if majuscules:
        grups_activats.append(grup_majuscules)

    if minuscules:
        grups_activats.append(grup_minuscules)

    if numeros:
        grups_activats.append(grup_numeros)

    if signes:
        grups_activats.append(grup_signes)

    if len(grups_activats) == 0:
        return "Error: has d'activar almenys un tipus de caràcter."

    if longitud < len(grups_activats):
        return "Error: la longitud és massa curta per incloure tots els tipus demanats."

    rng = np.random.default_rng()

    contrasenya = []

    for grup in grups_activats:
        contrasenya.append(rng.choice(list(grup)))

    tots_els_caracters = ""

    for grup in grups_activats:
        tots_els_caracters += grup

    while len(contrasenya) < longitud:
        contrasenya.append(rng.choice(list(tots_els_caracters)))

    rng.shuffle(contrasenya)

    return "".join(contrasenya)
```

```python
import pyperclip

contrasenya = crear_contrasenya(10, True, True, True, True)
contrasenya_2 = crear_contrasenya(10, True, True, True, True)
contrasenya_3 = crear_contrasenya(10, True, True, True, True)

pyperclip.copy(contrasenya)

print("La contrasenya s'ha copiat al porta-retalls.")
print(contrasenya)
print(contrasenya_2)
print(contrasenya_3)
```

**2. Processament de dades simple**

Una companya de feina ens ha demanat un favor, aprofitant que sap que estem aprenent a programar. Té un històric de partits de futbol català en un fitxer, on hi ha emmagatzemat els noms dels equips i els resultats. Necessita que processem les dades de manera automàtica, per extreure'n els resultats que necessita.


Utilitza el fitxer `"historic_partits.txt"`

Necessita un programa que retorni:

- El nombre total de gols que ha fet cada equip.
- El nom de l'equip més golejador.
- El nom de l'equip més golejat.
- La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts).


```python
def processar_partits():
    gols_fets = {}
    gols_rebuts = {}
    punts = {}

    with open("historic_partits.txt", "r", encoding="utf-8") as fitxer:
        for linia in fitxer:
            equip1, resultat, equip2 = linia.strip().split("\t")

            gols1, gols2 = resultat.split("-")
            gols1 = int(gols1)
            gols2 = int(gols2)

            if equip1 not in punts:
                punts[equip1] = 0
                gols_fets[equip1] = 0
                gols_rebuts[equip1] = 0

            if equip2 not in punts:
                punts[equip2] = 0
                gols_fets[equip2] = 0
                gols_rebuts[equip2] = 0

            gols_fets[equip1] += gols1
            gols_fets[equip2] += gols2

            gols_rebuts[equip1] += gols2
            gols_rebuts[equip2] += gols1

            if gols1 > gols2:
                punts[equip1] += 3
            elif gols2 > gols1:
                punts[equip2] += 3
            else:
                punts[equip1] += 1
                punts[equip2] += 1

    print("GOLS FETS PER CADA EQUIP")
    for equip in gols_fets:
        print(equip, ":", gols_fets[equip])

    print("\nEQUIP MÉS GOLEJADOR")
    equip_mes_golejador = max(gols_fets, key=gols_fets.get)
    print(equip_mes_golejador, "amb", gols_fets[equip_mes_golejador], "gols")

    print("\nEQUIP MÉS GOLEJAT")
    equip_mes_golejat = max(gols_rebuts, key=gols_rebuts.get)
    print(equip_mes_golejat, "amb", gols_rebuts[equip_mes_golejat], "gols rebuts")

    print("\nCLASSIFICACIÓ GLOBAL")
    classificacio = sorted(punts.items(), key=lambda x: x[1], reverse=True)

    for equip, punts_equip in classificacio:
        print(equip, ":", punts_equip, "punts")
```

```python
processar_partits()
```
