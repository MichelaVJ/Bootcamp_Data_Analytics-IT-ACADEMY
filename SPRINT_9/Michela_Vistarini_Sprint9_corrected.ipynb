**1. Calculadora de l'índex de massa corporal**

Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en: [🔗 Índice de masa corporal IMC que es y como se calcula.](https://muysalud.com/salud/indice-de-masa-corporal-imc-que-es-y-como-se-calcula/)

La funció ha de classificar el resultat en les seves respectives categories.

*Consell: Intenta validar les dades prèviament, perquè enviï un missatge d'advertència si les dades introduïdes per l'usuari/a no estan en el format adequat o no pren valors raonables.*


```python
def demanar_dada(missatge: str) -> float:
    try:
        return float(input(missatge).replace(",", "."))
    except ValueError:
        raise TypeError("Has d'introduir números.")


def validar_numero_positiu(valor: float, nom: str) -> None:
    if valor <= 0:
        raise ValueError(f"El {nom} ha de ser un valor positiu.")


def calcular_imc(pes: float, altura: float) -> float:
    return round(pes / (altura ** 2), 2)


def clasificar_imc(imc: float) -> str:
    if imc < 18.5:
        return "Pes baix"
    elif imc < 25:
        return "Pes normal"
    elif imc < 30:
        return "Sobrepès"
    else:
        return "Obesitat"


def mostrar_resultat(imc: float, categoria_pes: str) -> None:
    print(f"IMC: {imc}\nCategoria: {categoria_pes}.")


def calculadora_imc() -> None:
    try:
        pes = demanar_dada("Introdueix el teu pes en kg (ex. 64): ")
        altura = demanar_dada("Introdueix la teva altura en metres (ex. 1.75): ")
        validar_numero_positiu(pes, "pes")
        validar_numero_positiu(altura, "altura")
        imc = calcular_imc(pes, altura)
        categoria_pes = clasificar_imc(imc)
        mostrar_resultat(imc, categoria_pes)

    except (TypeError, ValueError) as error:
        print(f"{type(error).__name__}: {error}")
```

```python
calculadora_imc()
```

**2. Convertidor de temperatures**

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).

Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera que es puguin guardar (recorda que un `print()` no es pot guardar mai).

*Consell: Intenta validar les dades prèviament, perquè enviï un missatge d'advertència si les dades introduïdes per l'usuari/a no estan en el format adequat.*

**(EXTRA):** Pensa una manera d'emmagatzemar totes les possibles conversions en un sol objecte (Llista? Diccionari? DataFrame?) en comptes d'escriure molts `if else` en funció de la temperatura d'origen i la temperatura de destí.


```python
CONVERSIONS = {
    "C": {
        "min": -273.15,
        "to_kelvin": lambda t: t + 273.15,
        "from_kelvin": lambda t: t - 273.15
    },
    "F": {
        "min": -459.67,
        "to_kelvin": lambda t: (t - 32) * 5/9 + 273.15,
        "from_kelvin": lambda t: (t - 273.15) * 9/5 + 32
    },
    "K": {
        "min": 0,
        "to_kelvin": lambda t: t,
        "from_kelvin": lambda t: t
    }
}


def validar_unitat(unitat: str) -> None:
    if unitat not in CONVERSIONS:
        raise ValueError("La unitat ha de ser C, F o K.")


def validar_temperatura(temperatura: float, unitat: str) -> None:
    if temperatura < CONVERSIONS[unitat]["min"]:
        raise ValueError("La temperatura introduïda no és físicament possible.")


def convertir(temperatura: float, unitat_origen: str) -> dict:
    kelvin = CONVERSIONS[unitat_origen]["to_kelvin"](temperatura)
    return {
        unitat: round(dades["from_kelvin"](kelvin), 2)
        for unitat, dades in CONVERSIONS.items()
        if unitat != unitat_origen
    }


def convertir_temperatura() -> dict:
    temperatura = demanar_dada("Introdueix la temperatura: ")
    unitat_origen = input("Introdueix la unitat d'origen (C, F o K): ").upper()
    validar_unitat(unitat_origen)
    validar_temperatura(temperatura, unitat_origen)
    return convertir(temperatura, unitat_origen)
```

```python
try:
    resultat_conversio = convertir_temperatura()
    print(resultat_conversio)
except (TypeError, ValueError) as error:
    print(f"{type(error).__name__}: {error}")
```

**3. Comptador de paraules d'un text**

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula. Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament.

**(EXTRA):** Quina és la llargada mitjana de les paraules del text que has escrit? "Hola com va?" hauria de retornar (4+3+2) / 3 = 3


```python
from collections import Counter
from statistics import mean
import re
from pprint import pprint


def validar_text(text: str) -> None:
    if not isinstance(text, str) or not text.strip():
        raise ValueError("Has d'introduir un text no buit.")


def comptar_paraules(text: str) -> dict:
    paraules = re.findall(r"\b\w+\b", text.lower())
    return dict(Counter(paraules))


def calcular_llargada_mitjana(recompte: dict) -> float:
    return round(mean(map(len, recompte.keys())), 2)


def mostrar_resultat_paraules(recompte: dict, mitjana: float) -> None:
    pprint(recompte)
    print(f"\nLlargada mitjana de les paraules: {mitjana}")


def comptador_de_paraules() -> tuple:
    try:
        text = input("Introdueix un text: ")
        validar_text(text)
        recompte = comptar_paraules(text)
        mitjana = calcular_llargada_mitjana(recompte)
        mostrar_resultat_paraules(recompte, mitjana)
        return recompte, mitjana
    except ValueError as error:
        print(f"ValueError: {error}")
        return {}, None
```

**4. Diccionari invers (amb possibilitat de duplicats)**

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors.

Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.


```python
def invertir_diccionari(diccionari):

    if type(diccionari) != dict:
        return "Avís: has d'introduir un diccionari."

    if len(diccionari) == 0:
        return "Avís: el diccionari no pot estar buit."

    diccionari_invertit = {}

    for clau, valor in diccionari.items():

        if valor not in diccionari_invertit:
            diccionari_invertit[valor] = clau

        else:
            print(f"Avís: el valor '{valor}' està repetit.")

            if type(diccionari_invertit[valor]) != list:
                diccionari_invertit[valor] = [diccionari_invertit[valor]]

            diccionari_invertit[valor].append(clau)

            print(f"Claus associades al valor repetit: {diccionari_invertit[valor]}")

    return diccionari_invertit
```

```python
d = {"gat": "animal", "gos": "animal", "rosa": "planta"}
invertir_diccionari(d)
```

**1. Comptador i endreçador de paraules d'un text**

El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z. Per exemple, per a l'arxiu "tu\_me\_quieres\_blanca.txt".


```python
def llegir_arxiu(nom_arxiu: str) -> str:
    try:
        with open(nom_arxiu, "r", encoding="utf-8") as arxiu:
            text = arxiu.read()
        if not text.strip():
            raise ValueError("L'arxiu està buit.")
        return text
    except FileNotFoundError:
        raise FileNotFoundError("No s'ha trobat l'arxiu.")


def ordenar_per_lletra(recompte: dict) -> dict:
    resultat = {}
    for lletra in "abcdefghijklmnopqrstuvwxyz":
        paraules_lletra = {p: recompte[p] for p in sorted(recompte) if p.startswith(lletra)}
        if paraules_lletra:
            resultat[lletra] = paraules_lletra
    return resultat


def comptar_i_ordenar_paraules(nom_arxiu: str) -> dict:
    text = llegir_arxiu(nom_arxiu)
    recompte = comptar_paraules(text)
    return ordenar_per_lletra(recompte)
```

```python
from pprint import pprint

resultat = comptar_i_ordenar_paraules("tu_me_quieres_blanca.txt")
pprint(resultat)
```

**2. Conversió de tipus de dades**

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client:


```python
def conversio(llista_dades):
    convertibles = []
    no_convertibles = []

    def revisar_element(element):
        if type(element) == list or type(element) == tuple:
            for valor in element:
                revisar_element(valor)
        else:
            try:
                numero = float(element)
                convertibles.append(numero)
            except Exception:
                no_convertibles.append(element)

    for element in llista_dades:
        revisar_element(element)

    return convertibles, no_convertibles
```

**1. Generador de Contrasenyes**

Explora el funcionament del mòdul random de la llibreria numpy. [🔗 Random module in NumPy](https://www.cybrosys.com/blog/random-module-in-numpy)

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d'aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:

- `longitud (int)`: Longitud de la contrasenya
- `majuscules (bool = True)`: Si hi ha d'aparèixer majúscules
- `minuscules (bool = True)`: Si hi ha d'aparèixer minúscules
- `numeros (bool = True)`: Si hi ha d'aparèixer números
- `signes (bool = False)`: Si hi ha d'aparèixer caràcters especials (,-$? o similars)


Així doncs, si executem la funció de la següent manera: `crear_contrasenya(10, True, True, True, True)`

Hauríem d'obtenir un output (que hem de poder guardar) de l'estil: `9Er,5Vn8P$`

Assegura't que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.

**(EXTRA)** Explora com podríem fer que la funció copiés la contrasenya automàticament al porta papers de l'ordinador (com si l'haguéssim seleccionat i fet ctrl+copy).


```python
import string
import numpy as np
import pyperclip


def crear_contrasenya(longitud: int, majuscules: bool = True, minuscules: bool = True, numeros: bool = True, signes: bool = False) -> None:
    grups_activats = []

    if majuscules:
        grups_activats.append(string.ascii_uppercase)
    if minuscules:
        grups_activats.append(string.ascii_lowercase)
    if numeros:
        grups_activats.append(string.digits)
    if signes:
        grups_activats.append(string.punctuation)

    if not grups_activats:
        raise ValueError("Has d'activar almenys un tipus de caràcter.")
    if longitud < len(grups_activats):
        raise ValueError("La longitud és massa curta per incloure tots els tipus demanats.")

    rng = np.random.default_rng()
    tots_els_caracters = "".join(grups_activats)

    contrasenya = [rng.choice(list(grup)) for grup in grups_activats]
    while len(contrasenya) < longitud:
        contrasenya.append(rng.choice(list(tots_els_caracters)))

    rng.shuffle(contrasenya)

    pyperclip.copy("".join(contrasenya))
    print("La contrasenya s'ha copiat al porta-retalls.")
```

**2. Processament de dades simple**

Una companya de feina ens ha demanat un favor, aprofitant que sap que estem aprenent a programar. Té un històric de partits de futbol català en un fitxer, on hi ha emmagatzemat els noms dels equips i els resultats. Necessita que processem les dades de manera automàtica, per extreure'n els resultats que necessita.


Utilitza el fitxer `"historic_partits.txt"`

Necessita un programa que retorni:

- El nombre total de gols que ha fet cada equip.
- El nom de l'equip més golejador.
- El nom de l'equip més golejat.
- La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts).


```python
def llegir_partits(nom_arxiu: str) -> tuple:
    gols_fets = {}
    gols_rebuts = {}
    punts = {}

    with open(nom_arxiu, "r", encoding="utf-8") as fitxer:
        for linia in fitxer:
            equip1, resultat, equip2 = linia.strip().split("\t")
            gols1, gols2 = map(int, resultat.split("-"))

            for equip in [equip1, equip2]:
                if equip not in punts:
                    punts[equip] = 0
                    gols_fets[equip] = 0
                    gols_rebuts[equip] = 0

            gols_fets[equip1] += gols1
            gols_fets[equip2] += gols2
            gols_rebuts[equip1] += gols2
            gols_rebuts[equip2] += gols1

            if gols1 > gols2:
                punts[equip1] += 3
            elif gols2 > gols1:
                punts[equip2] += 3
            else:
                punts[equip1] += 1
                punts[equip2] += 1

    return gols_fets, gols_rebuts, punts


def mostrar_gols(gols_fets: dict) -> None:
    print("GOLS FETS PER CADA EQUIP")
    for equip, gols in gols_fets.items():
        print(f"  {equip}: {gols}")


def mostrar_mes_golejador(gols_fets: dict) -> None:
    equip = max(gols_fets, key=gols_fets.get)
    print(f"\nEQUIP MÉS GOLEJADOR\n  {equip} amb {gols_fets[equip]} gols")


def mostrar_mes_golejat(gols_rebuts: dict) -> None:
    equip = max(gols_rebuts, key=gols_rebuts.get)
    print(f"\nEQUIP MÉS GOLEJAT\n  {equip} amb {gols_rebuts[equip]} gols rebuts")


def mostrar_classificacio(punts: dict, gols_fets: dict, gols_rebuts: dict) -> None:
    print("\nCLASSIFICACIÓ GLOBAL")
    classificacio = sorted(
        punts.items(),
        key=lambda x: (x[1], gols_fets[x[0]] - gols_rebuts[x[0]]),
        reverse=True
    )
    for equip, pts in classificacio:
        diferencia = gols_fets[equip] - gols_rebuts[equip]
        print(f"  {equip}: {pts} pts (dif. gols: {diferencia:+})")


def processar_partits(nom_arxiu: str) -> None:
    gols_fets, gols_rebuts, punts = llegir_partits(nom_arxiu)
    mostrar_gols(gols_fets)
    mostrar_mes_golejador(gols_fets)
    mostrar_mes_golejat(gols_rebuts)
    mostrar_classificacio(punts, gols_fets, gols_rebuts)
```

```python
processar_partits("historic_partits.txt")
```
